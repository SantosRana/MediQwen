## Import all components

```python
from agent.state import AgentState
from agent.nodes import (
    run_dialogue_manager,
    run_risk_classification,
    run_multimodal_processing,
    run_vector_search,
    run_web_search_tool,
    safe_refusal_node,
    run_qwen_generation
)

def create_test_state(**kwargs) -> AgentState:
    """Creates a fully populated AgentState baseline payload."""
    base: AgentState = {
        "user_query": "",
        "image_path": None,
        "process_image": False,
        "processed_image_payload": None,
        "is_online": False,
        "dialog_state": "chat",
        "requires_retrieval": False,
        "clinical_subject": None,
        "followup_pending": False,
        "scope": "MEDICAL",
        "scope_source": "BGE_CLASSIFIER",
        "decision": "HIGH_CONFIDENCE",
        "risk_level": "LOW",
        "risk_metadata": {},
        "is_safe": True,
        "intent": "general",
        "top_retrieval_distance": 0.0,
        "retrieved_context": [],
        "context_sources": [],
        "retrieval_query": None,
        "agent_response": "",
        "routing_trace": [],
        "show_risk_badge": False
    }
    base.update(kwargs)
    return base
```

## Node 1: Dialogue Manager

```python
print("=" * 70)
print("🩺 DEMONSTRATION 1 [Node 1]: Dialogue Manager FSM & Intent Routing")
print("=" * 70)

# ---------------------------------------------------------------------
# Scenario 1A: Explicit Clinical Inquiry (Establishes Subject Memory)
# ---------------------------------------------------------------------
print("\n--- [Scenario 1A: Explicit Clinical Inquiry] ---")
state_1a = create_test_state(user_query="What are the main symptoms of urticaria?")
print("Query: ", state_1a.get("user_query"))
out_1a = run_dialogue_manager(state_1a)

print("📥 Output Inspection:")
print(f"  • BGE Scope             : {out_1a.get('scope')} (Source: {out_1a.get('scope_source')})")
print(f"  • Extracted Subject     : '{out_1a.get('clinical_subject')}'")
print(f"  • Dialogue State        : {out_1a.get('dialog_state')}")
print(f"  • Retrieval Authorized? : {out_1a.get('requires_retrieval')}")
print(f"  • Safety Flag (is_safe) : {out_1a.get('is_safe')}")

# ---------------------------------------------------------------------
# Scenario 1B: Anaphoric Follow-up (Context Resolution)
# ---------------------------------------------------------------------
print("\n--- [Scenario 1B: Contextual Anaphoric Follow-up] ---")
state_1b = create_test_state(
    user_query="How can I treat it at home?",
    clinical_subject=out_1a.get("clinical_subject")  # Pass 'urticaria' from Turn 1
)
print("Query: ", state_1b.get("user_query"))
out_1b = run_dialogue_manager(state_1b)

print("📥 Output Inspection:")
print(f"  • Incoming Active Subject: '{state_1b.get('clinical_subject')}'")
print(f"  • Preserved Subject     : '{out_1b.get('clinical_subject')}'")
print(f"  • Dialogue State        : {out_1b.get('dialog_state')}")
print(f"  • Retrieval Authorized? : {out_1b.get('requires_retrieval')}")

# ---------------------------------------------------------------------
# Scenario 1C: Conversational Casual Chat (No RAG Authorization)
# ---------------------------------------------------------------------
print("\n--- [Scenario 1C: Casual Conversation / Greeting] ---")
state_1c = create_test_state(user_query="Hello! Good morning, how are you today?")
print("Query: ", state_1c.get("user_query"))
out_1c = run_dialogue_manager(state_1c)

print("📥 Output Inspection:")
print(f"  • BGE Scope             : {out_1c.get('scope')}")
print(f"  • Dialogue State        : {out_1c.get('dialog_state')}")
print(f"  • Retrieval Authorized? : {out_1c.get('requires_retrieval')}")

# ---------------------------------------------------------------------
# Scenario 1D: Out-Of-Scope Non-Medical Query (Memory Pollution Shield)
# ---------------------------------------------------------------------
print("\n--- [Scenario 1D: Out-Of-Scope Query (Prevents State Corruption)] ---")
state_1d = create_test_state(user_query="What is machine learning?")
print("Query: ", state_1d.get("user_query"))
out_1d = run_dialogue_manager(state_1d)

print("📥 Output Inspection:")
print(f"  • BGE Scope             : {out_1d.get('scope')}")
print(f"  • Extracted Subject     : {out_1d.get('clinical_subject')} (Guarded from state pollution)")
print(f"  • Dialogue State        : {out_1d.get('dialog_state')}")
print(f"  • Retrieval Authorized? : {out_1d.get('requires_retrieval')}")

# ---------------------------------------------------------------------
# Scenario 1E: Technical / Code Generation Intercept
# ---------------------------------------------------------------------
print("\n--- [Scenario 1E: Malicious Code / Technical Request Intercept] ---")
state_1e = create_test_state(user_query="Write a python script to parse fasta medical sequences")
print("Query: ", state_1e.get("user_query"))
out_1e = run_dialogue_manager(state_1e)

print("📥 Output Inspection:")
print(f"  • Safety Flag (is_safe) : {out_1e.get('is_safe')}")
print(f"  • Dialogue State        : {out_1e.get('dialog_state')}")
print(f"  • Intercept Response    : '{out_1e.get('agent_response')}'")
```

# Node 3: Multimodal Processing

```python
print("=" * 70)
print("📸 DEMONSTRATION 3 [Node 3]: Multimodal Image Processing")
print("=" * 70)

# Scenario: Incoming turn containing a raw skin rash image asset
HIVES_IMAGE_PATH = PROJECT_ROOT / "data" / "test_assets" / "hives.jpg"

demo_state_multi = create_test_state(
    user_query="What is this on my arm?",
    image_path=HIVES_IMAGE_PATH,
    process_image=True,
    dialog_state="multimodal_triage"
)

multi_output = run_multimodal_processing(demo_state_multi)

payload = multi_output.get("processed_image_payload")

print("\n📥 Output Payload Inspection from run_multimodal_processing:")
print(f"  • Process Image Flag     : {demo_state_multi.get('process_image')}")
print(f"  • Payload Generated?     : {payload is not None}")
print(f"  • Payload Type           : {type(payload)}")

if payload:
    if isinstance(payload, str):
        print(f"  • Base64 Payload Length  : {len(payload):,} characters")
    elif isinstance(payload, bytes):
        print(f"  • Payload Size (KB)      : {len(payload) / 1024:.2f} KB")
    elif isinstance(payload, dict):
        print(f"  • Payload Metadata Keys  : {list(payload.keys())}")
        
# Compact Validation Assertions
assert demo_state_multi["process_image"] is True
assert payload is not None
assert isinstance(payload, str)

print("\n✅ Multimodal image preprocessing completed successfully!")
```

# Node 4: Local Vector Retrieval

```python
print("=" * 70)
print("🔍 DEMONSTRATION 4 [Node 4]: Local Vector Retrieval")
print("=" * 70)

# ---------------------------------------------------------------------
# Case 4A: Local Knowledge Base Hit (High Confidence Match)
# ---------------------------------------------------------------------
print("\n--- [Case 4A: Local KB Hit — Asthma Management] ---")
state_4a = create_test_state(
    user_query="What should I do during an asthma attack?",
    requires_retrieval=True,
    clinical_subject="asthma",
    risk_level="LOW"
)

out_4a = run_vector_search(state_4a)

print("📥 Output Payload Inspection:")
print(f"  • Resolved Search Query : '{out_4a.get('retrieval_query')}'")
print(f"  • Top Candidate Distance: {out_4a.get('top_retrieval_distance')}")
print(f"  • Chunks Retained       : {len(out_4a.get('retrieved_context', []))}")
print(f"  • Context Sources       : {out_4a.get('context_sources')}")

assert len(out_4a.get("retrieved_context", [])) > 0
assert "retrieved_context" in out_4a
print("✅ Case 4A Passed: Local DB candidate located.")

# ---------------------------------------------------------------------
# Case 4B: Local Knowledge Base Miss (Triggers Fallback Decision)
# ---------------------------------------------------------------------
print("\n--- [Case 4B: Local KB Miss — Unindexed Rare Condition] ---")
state_4b = create_test_state(
    user_query="How is Kawasaki disease diagnosed?",
    requires_retrieval=True,
    clinical_subject="Kawasaki disease",
    risk_level="LOW"
)

out_4b = run_vector_search(state_4b)

print("📥 Output Payload Inspection:")
print(f"  • Resolved Search Query : '{out_4b.get('retrieval_query')}'")
print(f"  • Chunks Retained       : {len(out_4b.get('retrieved_context', []))}")

assert len(out_4b.get("retrieved_context", [])) == 0
print("✅ Case 4B Passed: Local DB miss verified (0 chunks returned).")
```

# Node 5: Live Web Search Fallback

```python
print("=" * 70)
print("🌐 DEMONSTRATION 5 [Node 5]: Live Web Search Fallback & Content Inspection")
print("=" * 70)

# ---------------------------------------------------------------------
# Case 5A: Clinical Query Intent Normalization & Web Content Inspection
# ---------------------------------------------------------------------
print("\n--- [Case 5A: Clinical Web Search — Urticaria] ---")
state_5a = create_test_state(
    user_query="how can we treat it at home?",
    requires_retrieval=True,
    clinical_subject="urticaria",  # Subject preserved from prior turn
    is_online=True,
    risk_level="LOW"
)

out_5a = run_web_search_tool(state_5a)

print("📥 Node Metadata Inspection:")
print(f"  • Raw Input Query       : '{state_5a['user_query']}'")
print(f"  • Preserved Subject     : '{state_5a['clinical_subject']}'")
print(f"  • Normalized Web Term   : '{out_5a.get('retrieval_query')}'")
print(f"  • Chunks Retained       : {len(out_5a.get('retrieved_context', []))}")
print(f"  • Interleaved Sources   : {out_5a.get('context_sources')}")

print("\n📄 RETRIEVED WEB CONTENT PAYLOAD (First 2 Chunks):")
print("=" * 60)
for idx, chunk_text in enumerate(out_5a.get("retrieved_context", [])[:2], 1):
    print(f"--- [Chunk {idx}] ---")
    print(chunk_text.strip())
    print()
print("=" * 60)

assert out_5a.get("retrieval_query") == "urticaria treatment"
assert len(out_5a.get("retrieved_context", [])) > 0
print("✅ Case 5A Passed: Web content retrieved and verified.")

# ---------------------------------------------------------------------
# Case 5B: Nutrition Scoped Search & Content Inspection
# ---------------------------------------------------------------------
print("\n--- [Case 5B: Nutrition Web Search — Low Sodium Diet] ---")
state_5b = create_test_state(
    user_query="What foods should I eat on a low sodium diet?",
    requires_retrieval=True,
    is_online=True,
    risk_level="LOW"
)

out_5b = run_web_search_tool(state_5b)

print("📥 Node Metadata Inspection:")
print(f"  • Normalized Search Query: '{out_5b.get('retrieval_query')}'")
print(f"  • Chunks Retained        : {len(out_5b.get('retrieved_context', []))}")
print(f"  • Interleaved Sources    : {out_5b.get('context_sources')}")
print(f"  • Show Risk Badge?       : {out_5b.get('show_risk_badge')}")

print("\n📄 RETRIEVED NUTRITION CONTENT PAYLOAD (First Chunk):")
print("=" * 60)
if out_5b.get("retrieved_context"):
    print(out_5b["retrieved_context"][0].strip())
print("=" * 60)

assert out_5b.get("show_risk_badge") is False  # Suppressed for nutrition
print("✅ Case 5B Passed: Nutrition web search content verified.")
```
